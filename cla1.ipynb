```python
import pandas as pd
import numpy as np

np.random.seed(42)

skills = [
    "Python",
    "JavaScript",
    "Java",
    "C++",
    "SQL",
    "HTML_CSS",
    "React",
    "NodeJS",
    "MongoDB",
    "Git",
    "Docker",
    "Kubernetes",
    "AWS",
    "Machine_Learning",
    "Data_Structures",
    "Algorithms",
    "Computer_Networks",
    "Cybersecurity",
    "Hadoop",
    "Spark"
]

curriculum_score = [
    8, 7, 8, 8, 9,
    7, 5, 5, 5, 6,
    3, 2, 3, 4, 9,
    9, 7, 4, 3, 3
]

industry_score = [
    9, 9, 8, 7, 9,
    8, 8, 8, 7, 9,
    8, 7, 8, 9, 8,
    9, 8, 8, 7, 7
]

student_proficiency = [
    7, 5, 6, 6, 7,
    5, 4, 4, 4, 6,
    3, 2, 3, 3, 7,
    7, 5, 4, 3, 3
]

industry_demand = [
    9, 9, 8, 7, 9,
    8, 8, 8, 7, 9,
    8, 7, 8, 9, 9,
    9, 8, 8, 7, 7
]

df = pd.DataFrame({
    "skill_id": [f"S{i:03d}" for i in range(1, 21)],
    "skill": skills,
    "curriculum_score": curriculum_score,
    "industry_score": industry_score,
    "student_proficiency": student_proficiency,
    "industry_demand": industry_demand
})

# Calculate skill gap
df["skill_gap"] = df["industry_score"] - df["curriculum_score"]

# Calculate proficiency gap
df["proficiency_gap"] = df["industry_demand"] - df["student_proficiency"]

# Target:
# 1 = High-priority skill gap
# 0 = Low/Moderate-priority skill gap
df["target"] = (
    (df["skill_gap"] >= 2) |
    (df["proficiency_gap"] >= 4)
).astype(int)

df
```

```python
import pandas as pd
import numpy as np

# Load original dataset
original_path = f"{project_dir}/data/original_dataset.csv"
df = pd.read_csv(original_path)

# Create a copy for feature engineering
feature_df = df.copy()

# Feature 1: Curriculum-Industry Gap
feature_df["skill_gap"] = (
    feature_df["industry_score"] -
    feature_df["curriculum_score"]
)

# Feature 2: Student Proficiency Gap
feature_df["proficiency_gap"] = (
    feature_df["industry_demand"] -
    feature_df["student_proficiency"]
)

# Feature 3: Industry Demand Ratio
feature_df["industry_demand_ratio"] = (
    feature_df["industry_demand"] / 10.0
)

# Feature 4: Curriculum Coverage Ratio
feature_df["curriculum_coverage_ratio"] = (
    feature_df["curriculum_score"] / 10.0
)

# Feature 5: Student Proficiency Ratio
feature_df["student_proficiency_ratio"] = (
    feature_df["student_proficiency"] / 10.0
)

# Feature 6: Priority Score
feature_df["priority_score"] = (
    0.4 * feature_df["skill_gap"] +
    0.6 * feature_df["proficiency_gap"]
)

# Timestamp required by Feast
feature_df["event_timestamp"] = pd.Timestamp.now(tz="UTC")

# Display the engineered dataset
display(feature_df)
```

```python
training_code = '''
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# --------------------------------------------------
# Load historical features retrieved through Feast
# --------------------------------------------------

data_path = "/content/MLOps-Feast-SkillGap/outputs/historical_features.csv"

df = pd.read_csv(data_path)

# --------------------------------------------------
# Select Feast features
# --------------------------------------------------

feature_columns = [
    "curriculum_score",
    "industry_score",
    "student_proficiency",
    "industry_demand",
    "skill_gap",
    "proficiency_gap",
    "industry_demand_ratio",
    "curriculum_coverage_ratio",
    "student_proficiency_ratio",
    "priority_score"
]

X = df[feature_columns]
y = df["target"]

# --------------------------------------------------
# Split dataset
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

# --------------------------------------------------
# Train model
# --------------------------------------------------

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

# --------------------------------------------------
# Evaluate model
# --------------------------------------------------

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print()
print("Model Accuracy:", round(accuracy * 100, 2), "%")
print()

print("Classification Report:")
print(classification_report(y_test, y_pred))

# --------------------------------------------------
# Save model accuracy
# --------------------------------------------------

accuracy_path = (
    "/content/MLOps-Feast-SkillGap/"
    "outputs/model_accuracy.txt"
)

with open(accuracy_path, "w") as f:
    f.write(f"Model Accuracy: {accuracy * 100:.2f}%\\n")

print("Accuracy saved to:")
print(accuracy_path)
'''

training_path = f"{project_dir}/src/train_model.py"

with open(training_path, "w") as f:
    f.write(training_code)

print("train_model.py created successfully.")
```

```python
training_code = '''
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# --------------------------------------------------
# Load historical features retrieved through Feast
# --------------------------------------------------

data_path = "/content/MLOps-Feast-SkillGap/outputs/historical_features.csv"

df = pd.read_csv(data_path)

# --------------------------------------------------
# Feature columns
# --------------------------------------------------

feature_columns = [
    "curriculum_score",
    "industry_score",
    "student_proficiency",
    "industry_demand",
    "skill_gap",
    "proficiency_gap",
    "industry_demand_ratio",
    "curriculum_coverage_ratio",
    "student_proficiency_ratio",
    "priority_score"
]

X = df[feature_columns]
y = df["target"]

# --------------------------------------------------
# Train/test split
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# --------------------------------------------------
# Train Random Forest
# --------------------------------------------------

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

# --------------------------------------------------
# Evaluation
# --------------------------------------------------

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print()
print("Model Accuracy:", round(accuracy * 100, 2), "%")
print()
print("Classification Report:")
print(classification_report(y_test, y_pred))

# --------------------------------------------------
# Save model
# --------------------------------------------------

model_path = (
    "/content/MLOps-Feast-SkillGap/"
    "outputs/skill_gap_model.joblib"
)

joblib.dump(model, model_path)

# --------------------------------------------------
# Save accuracy
# --------------------------------------------------

accuracy_path = (
    "/content/MLOps-Feast-SkillGap/"
    "outputs/model_accuracy.txt"
)

with open(accuracy_path, "w") as f:
    f.write(f"Model Accuracy: {accuracy * 100:.2f}%\\n")

print("Model saved to:")
print(model_path)
'''

with open(training_path, "w") as f:
    f.write(training_code)

print("train_model.py updated successfully.")
```
